1. Count up with a condition


```python
i = 0
while i < 5:
    print(i)
    i += 1
```

2. Count down


```python
n = 3
while n > 0:
    print(n)
    n -= 1
```

3. Sum numbers from a list (index while)


```python
nums = [2, 5, 1]
i = 0
total = 0
while i < len(nums):
    total += nums[i]
    i += 1
print(total)
```

4. Stop at a sentinel value


```python
data = [1, 4, -1, 9]
i = 0
while i < len(data) and data[i] != -1:
    print(data[i])
    i += 1
```

5. break when target found


```python
items = [3, 8, 2, 5]
i = 0
target = 2
while i < len(items):
    if items[i] == target:
        print("found at", i)
        break
    i += 1
```

6. continue — skip even numbers


```python
n = 0
while n < 6:
    n += 1
    if n % 2 == 0:
        continue
    print(n)
```

7. while … else (no break)


```python
n = 2
while n > 0:
    print(n)
    n -= 1
else:
    print("finished normally")
```

8. while … else skipped by break


```python
n = 5
while n > 0:
    print(n)
    if n == 3:
        break
    n -= 1
else:
    print("done")
```

9. Menu-style while True + break


```python
rounds = 0
while True:
    rounds += 1
    print("round", rounds)
    if rounds >= 2:
        break
```

10. Count digits in a positive integer


```python
n = 408
count = 0
while n > 0:
    n //= 10
    count += 1
print(count)
```

11. Reverse digits of a number


```python
n = 123
rev = 0
while n > 0:
    rev = rev * 10 + n % 10
    n //= 10
print(rev)
```

12. Factorial with while


```python
n = 4
result = 1
while n > 1:
    result *= n
    n -= 1
print(result)
```

13. Fibonacci until value exceeds a limit


```python
limit = 20
a, b = 0, 1
while a <= limit:
    print(a)
    a, b = b, a + b
```

14. GCD (Euclidean algorithm)


```python
a, b = 48, 18
while b != 0:
    a, b = b, a % b
print(a)
```

15. Integer power base ** exp using multiplication


```python
base, exp = 3, 4
result = 1
while exp > 0:
    result *= base
    exp -= 1
print(result)
```

16. Collatz — print sequence until 1


```python
n = 6
while n != 1:
    print(n)
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
print(n)
```

17. Sum of digits

```python
n = 502
s = 0
while n > 0:
    s += n % 10
    n //= 10
print(s)
```

18. Check prime (trial division with while)


```python
n = 17
d = 2
is_prime = n >= 2
while d * d <= n and is_prime:
    if n % d == 0:
        is_prime = False
    d += 1
print(is_prime)
```

19. Count how many times you can halve an even number


```python
n = 40
steps = 0
while n % 2 == 0 and n > 0:
    n //= 2
    steps += 1
print(steps, n)
```

20. Double until exceeding a threshold


```python
x = 1
while x <= 30:
    print(x)
    x *= 2
```

21. Linear search with while (not found case)


```python
items = [1, 3, 5]
target = 4
i = 0
while i < len(items) and items[i] != target:
    i += 1
if i < len(items):
    print("at", i)
else:
    print("missing")
```

22. String traversal by index


```python
s = "go"
i = 0
out = ""
while i < len(s):
    out = s[i] + out
    i += 1
print(out)
```

23. Strip trailing spaces (concept)


```python
text = "hi   "
while text.endswith(" "):
    text = text[:-1]
print(repr(text))
```

24. First index where predicate holds


```python
nums = [3, -1, 4]
i = 0
while i < len(nums) and nums[i] >= 0:
    i += 1
if i < len(nums):
    print("first negative at", i)
else:
    print("none")
```

25. Reduce number by subtracting until in range


```python
x = 23
while x > 10:
    x -= 5
print(x)
```

26. Accumulate product until product > 50


```python
k = 1
p = 1
while p <= 50:
    p *= k
    k += 1
print(p)
```

27. Two-pointer merge (print pairs until one list ends)


```python
a = [1, 4]
b = [2, 3, 9]
i, j = 0, 0
while i < len(a) and j < len(b):
    if a[i] <= b[j]:
        print(a[i])
        i += 1
    else:
        print(b[j])
        j += 1
```

28. Palindrome check (characters)


```python
s = "racecar"
left, right = 0, len(s) - 1
ok = True
while left < right:
    if s[left] != s[right]:
        ok = False
        break
    left += 1
    right -= 1
print(ok)
```

29. Integer logarithm base 2 (floor)


```python
n = 31
count = -1
t = n
while t > 0:
    t //= 2
    count += 1
print(count)
```

30. Retry counter (max attempts)


```python
attempt = 0
max_try = 3
secret_ok = False
while attempt < max_try and not secret_ok:
    attempt += 1
    print("try", attempt)
    if attempt == 2:
        secret_ok = True
print("ok" if secret_ok else "fail")
```
