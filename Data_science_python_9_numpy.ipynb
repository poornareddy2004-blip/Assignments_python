```python
# Question 79: Top Three Store-Category Sales.

import numpy as np

store_names = np.array(["Store A", "Store B", "Store C"])
category_names = np.array(["Electronics", "Groceries", "Clothing"])

sales_matrix = np.array([
    [500, 700, 300],   # Store A
    [900, 200, 600],   # Store B
    [400, 800, 1000]   # Store C
])

flat_sales = sales_matrix.flatten()

sorted_indices = np.argsort(flat_sales)

top_3_indices = sorted_indices[-3:][::-1]

top_3_positions = np.unravel_index(top_3_indices, sales_matrix.shape)

print("Top 3 Store-Category Sales:")
for rank, (row, col) in enumerate(zip(*top_3_positions), start=1):
    value = sales_matrix[row, col]
    print(f"Rank {rank}: {value} | Store: {store_names[row]} | Category: {category_names[col]}")
```

```python
# Question 93: Replace with Company Average.

import numpy as np

avg_value = np.nanmean(sales_data)
sales_data_filled = np.where(np.isnan(sales_data), avg_value, sales_data)

print("Company Average (ignoring NaNs):", avg_value)
print("Sales Data with Missing Values Replaced:\n", sales_data_filled)
```

```python
# Question 94: Replace with Category Average.

import numpy as np

category_avg = np.nanmean(sales_data, axis=0)
inds = np.where(np.isnan(sales_data))
sales_data_filled = sales_data.copy()
sales_data_filled[inds] = np.take(category_avg, inds[1])

print("Category Averages:", category_avg)
print("Sales Data with Missing Values Replaced:\n", sales_data_filled)
```
