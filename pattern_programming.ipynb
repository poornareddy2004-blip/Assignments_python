1. Right-angled triangle — stars


```python
n = 4
for i in range(1, n + 1):
    print("*" * i)
```

2. Inverted right-angled triangle — stars


```python
n = 4
for i in range(n, 0, -1):
    print("*" * i)
```

3. Solid square of stars


```python
n = 3
for _ in range(n):
    print("*" * n)
```

4. Hollow square


```python
n = 5
for i in range(n):
    if i == 0 or i == n - 1:
        print("*" * n)
    else:
        print("*" + " " * (n - 2) + "*")
```

5. Number triangle (1, 12, 123, …)


```python
n = 4
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

6. Floyd’s triangle (continuous numbers)


```python
n = 4
num = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(num, end=" ")
        num += 1
    print()
```

7. Centered star pyramid


```python
n = 4
for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

8. Inverted centered star pyramid


```python
n = 4
for i in range(n, 0, -1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
```

9. Diamond (centered, stars)


```python
n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

10. Same digit per row


```python
n = 4
for i in range(1, n + 1):
    print((str(i) + " ") * i)
```

11. Binary triangle (0 and 1)


```python
n = 5
val = 1
for i in range(1, n + 1):
    for _ in range(i):
        print(val, end="")
        val = 1 - val
    print()
```

12. Checkerboard


```python
size = 4
for r in range(size):
    for c in range(size):
        cell = "#" if (r + c) % 2 == 0 else "."
        print(cell, end=" ")
    print()
```

13. Hollow diamond


```python
n = 4
for i in range(n):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")
for i in range(n - 2, -1, -1):
    left = n - i - 1
    mid = 2 * i - 1
    if i == 0:
        print(" " * left + "*")
    else:
        print(" " * left + "*" + " " * mid + "*")
```

14. Alphabet triangle


```python
n = 4
for i in range(1, n + 1):
    for j in range(i):
        print(chr(ord("A") + j), end="")
    print()
```

15. Centered number pyramid


```python
n = 4
for i in range(1, n + 1):
    nums = "".join(str(x) for x in range(1, i + 1))
    line = nums + nums[-2::-1] if i > 1 else nums
    pad = " " * (n - i)
    print(pad + line)
```

16. Right-aligned number triangle


```python
n = 4
for i in range(1, n + 1):
    nums = "".join(str(x) for x in range(1, i + 1))
    print(nums.rjust(n + i - 1))
```

17. Butterfly (stars, two wings)


```python
n = 3
for i in range(1, n + 1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
for i in range(n, 0, -1):
    print("*" * i + " " * (2 * (n - i)) + "*" * i)
```

18. Hourglass (stars)


```python
n = 5
for i in range(n, 0, -2):
    print(" " * ((n - i) // 2) + "*" * i)
for i in range(3, n + 1, 2):
    print(" " * ((n - i) // 2) + "*" * i)
```

19. Pascal’s triangle (coefficients)


```python
rows = 5
prev = []
for r in range(rows):
    cur = [1]
    for j in range(1, r):
        cur.append(prev[j - 1] + prev[j])
    if r > 0:
        cur.append(1)
    print(" ".join(str(x) for x in cur).center(rows * 3))
    prev = cur
```

20. X pattern (diagonals)


```python
n = 5
for r in range(n):
    for c in range(n):
        if r == c or r + c == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

21. Plus pattern (cross)


```python
n = 5
mid = n // 2
for r in range(n):
    for c in range(n):
        if r == mid or c == mid:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

22. Number grid (row × column values)


```python
n = 4
for r in range(1, n + 1):
    for c in range(1, n + 1):
        print(r * c, end="\t")
    print()
```

23. Hollow rectangle


```python
rows, cols = 3, 6
for r in range(rows):
    if r == 0 or r == rows - 1:
        print("*" * cols)
    else:
        print("*" + " " * (cols - 2) + "*")
```

24. Inverted number triangle (all digits to n)


```python
n = 4
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

25. Zigzag (stars on a shifting diagonal)


```python
cols = 7
for r in range(3):
    line = [" "] * cols
    for c in range(cols):
        if (c + r) % 3 == 0:
            line[c] = "*"
    print("".join(line))
```

26. GCD stripes (decorative bands)


```python
h, w = 4, 12
for r in range(h):
    for c in range(w):
        print("=" if (r + c) % 3 == 0 else "-", end="")
    print()
```

27. Staircase blocks (#)


```python
n = 4
for i in range(1, n + 1):
    print(("#" * i).ljust(n) + " " + ("#" * i))
```

28. Descending triangle of consecutive numbers


```python
n = 4
start = n * (n + 1) // 2
for i in range(n, 0, -1):
    row = []
    for _ in range(i):
        row.append(str(start))
        start -= 1
    print(" ".join(row))
```

29. Square border with digits (interior empty)


```python
n = 5
for r in range(n):
    for c in range(n):
        if r == 0 or r == n - 1 or c == 0 or c == n - 1:
            print((r + c) % 10, end="")
        else:
            print(" ", end="")
    print()
```

30. Arrow head (right-pointing)


```python
n = 5
for i in range(1, n + 1, 2):
    print("*" * i)
for i in range(n - 2, 0, -2):
    print("*" * i)
```
