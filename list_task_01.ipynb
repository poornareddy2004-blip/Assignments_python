# **LIST OPERATION**

```python
# Create a list of 5 employee names and add a new employee using append()
emplyoees=["asha","ravi","kiran","meena","ramya"]
emplyoees.append("suresh")
print(emplyoees)
```

```python
# Create a list of account numbers and add three new account numbers using extend().
account_numbers = ["c101","c102","c103","c104"]
new_account_numbers = ["c105","c106","c107"]
print(account_numbers)
account_numbers.extend(new_account_numbers)
print(account_numbers)
```

```python
# Insert a new customer at the beginning of a customer list using insert().
customer_list = ["c101","c1021","c103","c104"]
customer_list.insert(0,"c105")
print(customer_list)
```

```python
# Remove the last item from a list using pop().
customer_list = ["c101","c1021","c103","c104"]
customer_list.pop()
print(customer_list)
```

```python
# Remove the customer "Rahul" from a customer list using remove().
ustomer_list = ["c101","c1021","c103","c104"]
customer_list.remove("c1021")
print(customer_list)
```

```python
# Count how many times the account type "Savings" appears in a list using count()
account_types = ["savings","current","loan","savings","current","loan"]
account_types.count("savings")
```

```python
# Find the index of "Current" in a list of account types using index().
account_types = ["savings","current","loan"]
account_types.index("current")
```

```python
# Reverse a list of transaction IDs using reverse().
transacyion_ids = ["t101","t102","t103","t104","t104"]
transacyion_ids.reverse()
print(transacyion_ids)
```

```python
# Sort a list of account balances in ascending order using sort().
accountv_balances = [10000,20000,30000,40000]
accountv_balances.sort()
print(accountv_balances)
```

```python
# Sort a list of customer names in descending order using sort().
customer_names = ["ravi","kiran","meena","ramya"]
customer_names.sort(reverse=True)
print(customer_names)
```

```python
# Make a copy of a list using copy() and verify that modifying the copied list does not affect the original list.
original_list = [1,2,3,4,5]
copied_list = original_list.copy()
copied_list.append(4)
print(original_list)
print(copied_list)
```

```python
# Clear all elements from a list using clear().
original_list = [1,2,3,4,5]
original_list.clear()
print(original_list)
```

```python
# Append a nested list containing a new customer's details to an existing customer list.
customer_list = ["c101","c1021","c103","c104"]
new_customer = ["c105"]
customer_list.append(new_customer)
print(customer_list)
```

```python
# Extend a list of branch names with another list of new branches.
branch_names = ["b1","b2","b3"]
new_branch_names = ["b4","b5","b6"]
branch_names.extend(new_branch_names)
print(branch_names)
```

```python
# Remove the first occurrence of the balance 50000 from a list.
account_balances = [10000,20000,30000,40000,50000]
account_balances.remove(50000)
print(account_balances)
```

```python
# Pop the element at index 2 from a list and print the removed value.
account_balances = [10000,20000,30000,40000,50000]
removed_value = account_balances.pop(2)
print(removed_value)
```

```python
# Count how many times the city "Bangalore" appears in a list.
cities = ["bangalore","delhi","mumbai","bangalore"]
cities.count("bangalore")
```

```python
# Reverse a list of bank names and print the result.
bank_names = ["bank1","bank2","bank3"]
bank_names.reverse()
print(bank_names)
```

```python
# Create a list of loan amounts and sort it in descending order.
loan_amounts = [100000,200000,300000]
loan_amounts.sort(reverse=True)
print(loan_amounts)
```
