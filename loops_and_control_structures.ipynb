Problem 1.1: Grade Calculator (Easy)

```python
marks = int(input("Enter student's marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks! Please enter marks between 0 and 100.")
elif marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Grade F")
```

Problem 1.2: Leap Year Checker (Medium)

```python
year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

Problem 1.3: Triangle Type Identifier (Medium)

```python
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b > c and b + c > a and a + c > b:
    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")
else:
    print("Invalid triangle")
```

Problem 2.2: Factorial Calculator (Medium)

```python
n = int(input("Enter a number: "))
result = 1

if n < 0:
    print("Factorial is not defined for negative numbers")
else:
    for i in range(1, n + 1):
        result = result * i

    print("Factorial =", result)
```

Problem 2.5: Prime Number Checker (Medium)

```python
n = int(input("Enter a number: "))

if n <= 1:
    print("Not a prime number")
else:
    prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Prime number")
    else:
        print("Not a prime number")
```

Section 3: While Loops

Problem 3.1: Reverse a Number (Medium)

```python
n = int(input("Enter a number: "))

num = abs(n)
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if n < 0:
    reverse = -reverse

print("Reversed number =", reverse)
```

Problem 3.2: Sum of Digits (Easy)

```python
n = int(input("Enter a number: "))

n = abs(n)
total = 0

while n > 0:
    digit = n % 10
    total += digit
    n = n // 10

print("Sum of digits =", total)
```

Problem 3.4: Palindrome Checker (Medium)

```python
num = int(input("Enter a number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a palindrome number")
else:
    print(original, "is not a palindrome number")
```

Problem 3.5: Armstrong Number (Hard)

```python
num = int(input("Enter a number: "))
original = num
digits = len(str(num))
total = 0

while num > 0:
    digit = num % 10
    total += digit ** digits
    num = num // 10

if total == original:
    print(original, "is an Armstrong number")
else:
    print(original, "is not an Armstrong number")
```

Section 4: Nested Loops

Problem 4.1: Pattern Printing - Right Triangle (Easy)

```python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end="")
    print()
```

Problem 4.3: Pattern Printing - Pyramid (Medium)

```python
n = 5

for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()
```

Problem 4.4: Pattern Printing - Number Pattern (Medium)

```python
n = 5

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

Problem 4.5: Multiplication Table Grid (Medium)

```python
n = 5

for i in range(1, n + 1):
    for j in range(1, n + 1):
        print(f"{i * j:4}", end="")
    print()
```

Problem 4.6: Diamond Pattern (Hard)

```python
n = 9

for i in range(1, (n + 1) // 2 + 1):
    print(" " * (((n + 1) // 2) - i) +
          "*" * (2 * i - 1))

for i in range((n + 1) // 2 - 1, 0, -1):
    print(" " * (((n + 1) // 2) - i) +
          "*" * (2 * i - 1))
```

Problem 5.2: Skip Multiples of 3

```python
for i in range(1, 51):
    if i % 3 == 0:
        continue
    print(i, end=" ")
```

```python
total = 0

while True:
    num = int(input("Enter a number: "))

    if num < 0:
        break
    if num == 0:
        continue

    total += num

print("Final sum:", total)
```
