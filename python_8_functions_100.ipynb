```python
# 3.Create four separate functions:

def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Error: Division by zero is not allowed."
    return a / b


first_number = float(input("Enter first number: "))
second_number = float(input("Enter second number: "))
operation = input("Enter operation: ").strip().lower()

if operation == "add":
    result = add(first_number, second_number)
elif operation == "subtract":
    result = subtract(first_number, second_number)
elif operation == "multiply":
    result = multiply(first_number, second_number)
elif operation == "divide":
    result = divide(first_number, second_number)
else:
    result = "Error: Invalid operation. Choose add, subtract, multiply, or divide."

print(f"Result: {result}")
```

```python
# 8.Area and Circumference of a Circle.

import math

def circle_area(radius):
    return math.pi * radius ** 2

def circle_circumference(radius):
    return 2 * math.pi * radius


radius = float(input("Enter radius: "))

area = circle_area(radius)
circumference = circle_circumference(radius)

print(f"Area: {round(area, 2)}")
print(f"Circumference: {round(circumference, 2)}")
```

```python
# Assignment 13: Factorial Using a Function.

def calculate_factorial(number):
    if number < 0:
        print("Error: Factorial is not defined for negative numbers.")
        return None

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i
    return factorial


number = int(input("Enter a number: "))
result = calculate_factorial(number)

if result is not None:
    print(f"Factorial: {result}")
```

```python
# Assignment 16: Reverse a String.

def reverse_string_slicing(text):

    return text[::-1]


def reverse_string_loop(text):

    reversed_text = ""
    for character in text:
        reversed_text = character + reversed_text
    return reversed_text


text = input("Enter text: ")

result_slicing = reverse_string_slicing(text)
print(f"Reversed text (slicing): {result_slicing}")

result_loop = reverse_string_loop(text)
print(f"Reversed text (loop): {result_loop}")
```

```python
# Assignment 19: Find Maximum in a List.
def find_list_maximum(numbers):
    if len(numbers) == 0:
        return "The list is empty."

    largest = numbers[0]
    for number in numbers:
        if number > largest:
            largest = number
    return largest


text_input = input("Enter numbers: ")
numbers = [float(n) if "." in n else int(n) for n in text_input.split()]

result = find_list_maximum(numbers)
print(f"Largest number: {result}")
```

```python
# Assignment 20: Count Positive, Negative and Zero Values.

def count_number_types(numbers):
    positive_count = 0
    negative_count = 0
    zero_count = 0

    for number in numbers:
        if number > 0:
            positive_count += 1
        elif number < 0:
            negative_count += 1
        else:
            zero_count += 1

    return positive_count, negative_count, zero_count


text_input = input("Enter integers: ")
numbers = [int(n) for n in text_input.split()]

positives, negatives, zeros = count_number_types(numbers)

print(f"Positive numbers: {positives}")
print(f"Negative numbers: {negatives}")
print(f"Zeros: {zeros}")
```

```python
# Assignment 32: Function Returning a Dictionary.

def analyze_text(text):
    vowels = "aeiouAEIOU"
    total_characters = len(text)
    total_words = len(text.split())
    total_vowels = sum(1 for character in text if character in vowels)
    total_digits = sum(1 for character in text if character.isdigit())
    total_spaces = sum(1 for character in text if character == " ")

    return {
        "Total characters": total_characters,
        "Total words": total_words,
        "Total vowels": total_vowels,
        "Total digits": total_digits,
        "Total spaces": total_spaces
    }


text = input("Enter text: ")
result = analyze_text(text)

for key, value in result.items():
    print(f"{key}: {value}")
```

```python
# Assignment 33: Function Returning a List.

def get_even_numbers(start, end):
    if start >= end:
        print("Error: Starting number must be smaller than ending number.")
        return None

    even_numbers = []
    for number in range(start, end + 1):
        if number % 2 == 0:
            even_numbers.append(number)
    return even_numbers


start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

result = get_even_numbers(start, end)

if result is not None:
    print(f"Even numbers: {result}")
```

```python
# Assignment 34: Prime Number Function.

def is_prime(number):
    if number < 2:
        return False

    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


print("Prime numbers from 1 to 100:")
prime_numbers = []
for number in range(1, 101):
    if is_prime(number):
        prime_numbers.append(number)

print(prime_numbers)
```

```python
# Assignment 35: Fibonacci Function.

def generate_fibonacci(number_of_terms):
    if number_of_terms <= 0:
        print("Error: Number of terms must be a positive integer.")
        return []

    if number_of_terms == 1:
        return [0]

    fibonacci_series = [0, 1]
    for i in range(2, number_of_terms):
        next_term = fibonacci_series[i - 1] + fibonacci_series[i - 2]
        fibonacci_series.append(next_term)

    return fibonacci_series


number_of_terms = int(input("Enter number of terms: "))
result = generate_fibonacci(number_of_terms)
print(result)
```

```python
# Assignment 67: Filter Prime Numbers.

def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


text_input = input("Enter integers: ")
numbers = [int(n) for n in text_input.split()]

prime_numbers = list(filter(is_prime, numbers))

print(f"Original numbers: {numbers}")
print(f"Prime numbers: {prime_numbers}")
```

```python
# Assignment 68: Filter Palindrome Words.

import string


def is_palindrome_word(word):
    cleaned_word = word.lower().strip(string.punctuation)
    return cleaned_word == cleaned_word[::-1] and len(cleaned_word) > 0


sentence = input("Enter a sentence: ")
words = sentence.split()

palindrome_words = list(filter(is_palindrome_word, words))

print(f"Words: {words}")
print(f"Palindrome words: {palindrome_words}")
```

```python
# Assignment 69: Filter Valid Email Addresses.

def is_valid_email(email):
    if email.count("@") != 1:
        return False

    username, domain = email.split("@")

    if len(username) == 0:
        return False

    if "." not in domain:
        return False

    return True


emails = ["bharath@gmail.com", "invalid-email.com", "@gmail.com", "kiran@site", "meena@yahoo.co.in"]

valid_emails = list(filter(is_valid_email, emails))

print(f"Emails: {emails}")
print(f"Valid emails: {valid_emails}")
```

```python
# Assignment 74: Filter Perfect Squares'

import math


def is_perfect_square(number):
    root = int(math.sqrt(number))
    return root * root == number


text_input = input("Enter positive integers: ")
numbers = [int(n) for n in text_input.split()]

perfect_squares = list(filter(is_perfect_square, numbers))

print(f"Input: {numbers}")
print(f"Output: {perfect_squares}")
```

```python
# Assignment 75: Filter Dates by Year.
from datetime import datetime

dates = ["15-08-2020", "26-01-2021", "02-10-2020", "invalid-date", "31-12-2022", "14-05-2020"]

year_to_find = int(input("Enter year: "))


def belongs_to_year(date_string):
    try:
        parsed_date = datetime.strptime(date_string, "%d-%m-%Y")
        return parsed_date.year == year_to_find
    except ValueError:
        return False


matching_dates = list(filter(belongs_to_year, dates))

print(f"Dates: {dates}")
print(f"Dates in {year_to_find}: {matching_dates}")
```

```python
# Assignment 81: Calculate Factorial Using reduce().

from functools import reduce


def calculate_factorial(number):
    if number < 0:
        print("Error: Factorial is not defined for negative numbers.")
        return None

    if number == 0:
        return 1

    return reduce(lambda accumulator, value: accumulator * value, range(1, number + 1))


number = int(input("Enter a number: "))
result = calculate_factorial(number)

if result is not None:
    print(f"Factorial: {result}")
```

```python
# Assignment 82: Calculate Total Cart Price.

from functools import reduce

cart = [
    {"product": "Laptop", "price": 70000, "quantity": 1},
    {"product": "Mouse", "price": 800, "quantity": 2},
    {"product": "Keyboard", "price": 1500, "quantity": 1}
]

# Use map() to calculate each item's total
item_totals = list(map(lambda item: item["price"] * item["quantity"], cart))

print("Item-wise totals:")
for item, total in zip(cart, item_totals):
    print(f"{item['product']}: {total}")

# Use reduce() to calculate the grand total
grand_total = reduce(lambda accumulator, value: accumulator + value, item_totals, 0)

print(f"\nGrand total: {grand_total}")
```

```python
# Assignment 85: Compose Map, Filter and Reduce.
from functools import reduce

text_input = input("Enter integers: ")
numbers = [int(n) for n in text_input.split()]

# Step 1: filter even numbers
even_numbers = list(filter(lambda n: n % 2 == 0, numbers))

# Step 2: square the selected numbers
squares = list(map(lambda n: n ** 2, even_numbers))

# Step 3: sum the squares
final_total = reduce(lambda accumulator, value: accumulator + value, squares, 0)

print(f"Input: {numbers}")
print(f"Even numbers: {even_numbers}")
print(f"Squares: {squares}")
print(f"Final total: {final_total}")
```

```python
# Assignment 90: Closure-Based Discount Calculator.

def create_discount(discount_percentage):
    def apply_discount(price):
        return price - (price * discount_percentage / 100)

    return apply_discount


discount_10 = create_discount(10)
discount_20 = create_discount(20)
discount_30 = create_discount(30)

price = 2000

print(f"Price after 10% discount: {discount_10(price)}")
print(f"Price after 20% discount: {discount_20(price)}")
print(f"Price after 30% discount: {discount_30(price)}")
```

```python
# Assignment 91: Closure-Based Counter'
def create_counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment


counter = create_counter()

for _ in range(5):
    print(counter())
```

```python
# Assignment 92: Closure-Based Bank Balance.
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        print(f"Deposited: {amount}. New balance: {balance}")

    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            print(f"Error: Insufficient balance. Available: {balance}")
        else:
            balance -= amount
            print(f"Withdrew: {amount}. New balance: {balance}")

    def check_balance():
        print(f"Current balance: {balance}")

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(500)
withdraw(300)
withdraw(5000)
check_balance()
```

```python
#Assignment 93: Function Decorator.
def execution_logger(func):
    def wrapper(*args, **kwargs):
        print("Starting function...")
        result = func(*args, **kwargs)
        print(f"Result: {result}")
        print("Function completed.")
        return result

    return wrapper


@execution_logger
def add_numbers(a, b):
    return a + b


final_result = add_numbers(10, 20)
```

```python
# Assignment 99: LEGB Name Resolution.
value = "global value"


def outer_function():
    value = "enclosing value"

    def inner_function():
        value = "local value"
        print(f"Inside inner_function(): {value}")

    print(f"Inside outer_function() (before calling inner): {value}")
    inner_function()


print(f"Global scope: {value}")
outer_function()

# --- Now remove the local declaration in inner_function ---
def outer_function_no_local():
    value = "enclosing value"

    def inner_function_no_local():
        # No local 'value' here - Python looks one level up (Enclosing)
        print(f"Inside inner_function_no_local(): {value}")

    inner_function_no_local()


print()
outer_function_no_local()
def outer_function_no_enclosing():
    def inner_function_no_enclosing():
        print(f"Inside inner_function_no_enclosing(): {value}")

    inner_function_no_enclosing()


print()
outer_function_no_enclosing()
```
